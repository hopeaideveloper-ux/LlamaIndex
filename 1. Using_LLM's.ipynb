**LlamaIndex is an LLM application framework that connects LLMs to external/private data and provides tools for retrieval, querying, agents, workflows, and multimodal applications.**

# USING LLM

```python
from dotenv import load_dotenv
from llama_index.llms.openai import OpenAI
load_dotenv()

llm = OpenAI(model="gpt-4o-mini", temperature=0)
```

**1. OPEN AI TEXT LLM**

```python
# Ask a question
response = llm.complete("Explain artificial intelligence in one simple sentence.")

# Print only the generated text
print("Answer:", response.text)
```

**2. OPEN AI TEXT LLM USING CHAT MESSAGE**

```python
from llama_index.core.llms import ChatMessage

messages = [
    ChatMessage(
        role="system",
        content="You are an AI teacher. Explain concepts simply."
    ),
    ChatMessage(
        role="user",
        content="What is machine learning?"
    )
]

response = llm.chat(messages)
print("Answer:", response.message.content)
```

**3. MULTIMODAL LLM USING TEXTBLOCK AND IMAGEBLOCK**

```python
from pathlib import Path
from llama_index.llms.openai import OpenAI
from llama_index.core.llms import (ChatMessage, TextBlock, ImageBlock)

image_path = Path("image1.png")

# Create a message containing text and image
message = ChatMessage(
    role="user",
    blocks=[
        TextBlock(
            text=(
                "Describe this product image in three simple "
                "sentences. Mention its color, type and visible features."
            )
        ),
        ImageBlock(
            path=str(image_path)
        )
    ]
)

# Send the multimodal message
response = llm.chat([message])

# Print only the answer
print("\nImage Description:")
print(response.message.content)
```
