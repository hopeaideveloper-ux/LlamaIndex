```python
from dotenv import load_dotenv
from llama_index.llms.openai import OpenAI

load_dotenv()

llm = OpenAI(model="gpt-4o-mini", temperature=0)
```

**PROMPT TEMPLATE - PromptTemplate is used for simple text prompts. Variables use single braces such as {topic}.**

```python
from llama_index.core import PromptTemplate

# Create the prompt template
prompt_template = PromptTemplate(
    """
    You are an AI teacher.

    Explain {topic} in two simple sentences.
    Give one real-world example.
    """
)

# Get input
topic = input("Enter a topic: ")

# Format the prompt
formatted_prompt = prompt_template.format(topic=topic)

# Send it to the LLM
response = llm.complete(formatted_prompt)

print("\nAnswer:")
print(response.text)
```

**RichPromptTemplate - RichPromptTemplate uses Jinja-style syntax.**<br>
**Variables: {{ variable }}**<br>
**Conditions: {% if condition %}**<br>
**Loops: {% for item in items %}**

```python
from llama_index.core.prompts import RichPromptTemplate

# Create a rich prompt
rich_template = RichPromptTemplate(
    """
    You are an AI teacher.

    Explain {{ topic }} for a {{ level }} student.

    {% if include_example %}
    Include one simple real-world example.
    {% endif %}

    Keep the answer within three sentences.
    """
)

# Get input
topic = input("Enter a topic: ")
level = input("Enter student level: ")

# Format the prompt
formatted_prompt = rich_template.format(
    topic=topic,
    level=level,
    include_example=True
)

# Send it to the LLM
response = llm.complete(formatted_prompt)

print("\nAnswer:")
print(response.text)
```

**RichPromptTemplate WITH A LOOP**

```python
from llama_index.core.prompts import RichPromptTemplate

rich_template = RichPromptTemplate(
    """
    Explain {{ topic }} using the following sections:

    {% for section in sections %}
    - {{ section }}
    {% endfor %}
    """
)

topic = input("Enter a topic: ")

formatted_prompt = rich_template.format(
    topic=topic,
    sections=[
        "Definition",
        "Real-world example",
        "Simple summary"
    ]
)

response = llm.complete(formatted_prompt)

print(response.text)
```

**ChatPromptTemplate - Creates multiple messages with different roles.**
**The main roles are:**<br>

**SYSTEM: defines the behaviour of the LLM**<br>
**USER: contains the user’s question**<br>
**ASSISTANT: represents an earlier assistant response**

```python
from llama_index.core import ChatPromptTemplate
from llama_index.core.llms import ChatMessage, MessageRole

# Define message templates
message_templates = [
    ChatMessage(
        role=MessageRole.SYSTEM,
        content=(
            "You are an AI teacher. "
            "Explain technical concepts using simple language."
        )
    ),
    ChatMessage(
        role=MessageRole.USER,
        content=(
            "Explain {topic} in two sentences "
            "and give one example."
        )
    )
]

# Create the chat prompt
chat_template = ChatPromptTemplate(message_templates=message_templates)

# Get input
topic = input("Enter a topic: ")

# Format the messages
messages = chat_template.format_messages(
    topic=topic
)

# Send the messages to the LLM
response = llm.chat(messages)

print("\nAnswer:")
print(response.message.content)
```
