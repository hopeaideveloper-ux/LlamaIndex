# A RESSPONSE SYNTHESIZER combines the user’s question with retrieved nodes to produce the final response.

<div align="center">
    <img src="images/responsesynt.png" width="300" height="700">
</div>

```python
import logging
import warnings

warnings.filterwarnings("ignore", message="Mixing V1 models and V2 models")
logging.getLogger("httpx").setLevel(logging.WARNING)
logging.getLogger("httpcore").setLevel(logging.WARNING)
logging.getLogger("llama_index").setLevel(logging.ERROR)
logging.disable(logging.WARNING)
```

```python
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.llms.openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

# Load PDF
documents = SimpleDirectoryReader(input_files=["data/HopeAI.pdf"]).load_data()

# Create index
index = VectorStoreIndex.from_documents(documents)

# Create LLM
llm = OpenAI(model="gpt-4o-mini")

# User question
question = "What courses are offered by HopeAI?"
```

<h1 style="color: #16A34A;">1. COMPACT</h1>

# Combines retrieved chunks efficiently and generates an answer.

<div align="center">
    <img src="images/compactresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="compact"
)
response = query_engine.query(question)
print(response)
```

<h1 style="color: #16A34A;">2. REFINE</h1>

# Creates an initial answer and then refines it using each retrieved node.

<div align="center">
    <img src="images/refineresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="refine"
)
response = query_engine.query(question)
print(response)
```

<h1 style="color: #16A34A;">3. TREE SUMMARIZE</h1>

# Summarizes nodes using a tree structure.

<div align="center">
    <img src="images/treeresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="tree_summarize"
)
response = query_engine.query("Summarize the information about HopeAI.")
print(response)
```

<h1 style="color: #16A34A;">4. SIMPLE SUMMARIZE</h1>

# Combines the retrieved nodes and makes one LLM call to summarize them.

<div align="center">
    <img src="images/simpleresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="simple_summarize"
)
response = query_engine.query("Give me a summary of HopeAI.")
print(response)
```

<h1 style="color: #16A34A;">5. ACCUMULATE</h1>

# Generates a separate response for each retrieved node.

<div align="center">
    <img src="images/accumulateresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="accumulate"
)
response = query_engine.query(question)
print(response)
```

<h1 style="color: #16A34A;">6. COMPACT ACCUMULATE</h1>

# First combines nodes into compact groups and then generates a separate answer for each group.

<div align="center">
    <img src="images/compactaccresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    llm=llm,
    response_mode="compact_accumulate"
)
response = query_engine.query(question)
print(response)
```

<h1 style="color: #16A34A;">7. CONTEXT ONLY</h1>

# Returns the retrieved context itself instead of asking the LLM to generate a final answer.

<div align="center">
    <img src="images/contextonlyresponse.png" width="300" height="700">
</div>

```python
query_engine = index.as_query_engine(
    response_mode="context_only"
)
response = query_engine.query(question)
print(response)
```

<h1 style="color: #16A34A;">8. NO TEXT</h1>

# Used when you don't want normal text synthesis and instead want access to the retrieved source nodes.

<div align="center">
    <img src="images/notextresponse.png" width="300" height="700">
</div>

```python
retriever = index.as_retriever()
nodes = retriever.retrieve(question)

for i, node in enumerate(nodes, 1):
    print(f"\nNode {i}")
    print(node.text)
```

<h1 style="color: #16A34A;">9. GENERATION</h1>

# Generates an answer using the LLM without using the retrieved context.

<div align="center">
    <img src="images/generationresponse.png" width="300" height="700">
</div>

```python
response = llm.complete("What are the benefits of learning Artificial Intelligence?")
print(response)
```
